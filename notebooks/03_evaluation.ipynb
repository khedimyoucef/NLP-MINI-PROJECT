Generate sample stories for each age prefix and compute readability scores to verify conditioning.

```python
from src.model.model import load_model
from src.evaluate import readability_score
model, tokenizer = load_model('gpt2')
prefixes = ['Level: Age3-4', 'Level: Age5-6', 'Level: Age7-8', 'Level: Age9-10', 'Level: Age11-12']
theme = 'a dragon afraid of fire'
for prefix in prefixes:
    prompt = prefix + ' ' + theme
    inputs = tokenizer(prompt, return_tensors='pt')
    out = model.generate(**inputs, max_new_tokens=120)
    text = tokenizer.decode(out[0], skip_special_tokens=True)
    scores = readability_score(text)
    print(prefix, 'FK grade:', scores['flesch_kincaid_grade'])
```
